```python
import math
import matplotlib.pyplot as plt
import numpy as np

def calculate_sphere_segment_height_and_depth(R, D):
    """
    Расчитывает:
    - d: расстояние от центра сферы до плоскости сечения
    - h: высоту шарового сегмента
    - delta: глубину шлифовки
    Возвращает (d, h, delta).
    """
    # Расстояние от центра сферы до плоскости сечения
    d = math.sqrt(R**2 - (D / 2)**2)
    # Высота шарового сегмента
    h = R - d
    # Глубина шлифованного слоя
    delta = R - h  # то же, что и d
    
    return d, h, delta

def visualize_sphere_segment(R, D, h, d):
    """
    Строит 2D-визуализацию:
    - Окружность радиуса R (вид сбоку).
    - Хорду длиной D на высоте y=c = sqrt(R^2 - (D/2)^2).
    - Заштрихованную область шарового сегмента.
    """
    # Координата по оси y, на которой расположена хорда
    c = math.sqrt(R**2 - (D / 2)**2)  # это и есть d
    
    # Подготовим координаты для окружности
    theta = np.linspace(0, 2 * np.pi, 300)
    x_circ = R * np.cos(theta)
    y_circ = R * np.sin(theta)
    
    # Построим дугу верхней половины окружности, чтобы "залить" сегмент
    x_fill = np.linspace(-D/2, D/2, 200)
    y_fill_upper = np.sqrt(R**2 - x_fill**2)  # верхняя часть окружности

    # Построение фигуры
    plt.figure(figsize=(6, 6))
    plt.title("Шаровой сегмент (2D-сечение)")
    plt.plot(x_circ, y_circ, 'k', label="Окружность (сечение сферы)")
    
    # Заштриховываем сегмент между y=c и верхней дугой
    plt.fill_between(x_fill, c, y_fill_upper, color='skyblue', alpha=0.5, label="Шаровой сегмент")

    # Хорда - красная линия
    plt.plot([-D/2, D/2], [c, c], 'r-', linewidth=2, label="Основание сегмента (диаметр D)")
    
    # Отметим центр сферы точкой
    plt.plot(0, 0, 'ko', label="Центр сферы (0,0)")
    
    # Настраиваем оси
    plt.axis('equal')  # одинаковый масштаб по осям
    plt.grid(True)
    
    # Границы, чуть больше радиуса
    pad = 0.1 * R
    plt.xlim(-R - pad, R + pad)
    plt.ylim(-R - pad, R + pad)
    
    plt.legend()
    plt.show()

def main():
    print("Программа для расчёта высоты шарового сегмента и глубины шлифовки.\n")
    
    # Ввод исходных данных
    R = float(input("Введите радиус сферы (мм или мкм): "))
    D = float(input("Введите диаметр основания шарового сегмента (мм или мкм): "))
    
    # Расчёт
    d, h, delta = calculate_sphere_segment_height_and_depth(R, D)
    
    # Вывод результатов
    print(f"\nРасстояние от центра сферы до плоскости сечения (d) = {d:.4f}")
    print(f"Высота шарового сегмента (h) = {h:.4f}")
    print(f"Глубина шлифовки (Δ = R - h) = {delta:.4f}")
    
    # Визуализация
    visualize_sphere_segment(R, D, h, d)

if __name__ == "__main__":
    main()
```

```python
import math
import matplotlib.pyplot as plt
import numpy as np

# --------------------------------
# 1. Задаём толщины 4 слоёв (константы).
#    Сумма этих толщин должна совпадать с радиусом сферы.
# --------------------------------
layer_thicknesses = [0.3, 0.2, 0.1, 0.4]  # Пример: в мм
R = sum(layer_thicknesses)               # Итоговый радиус сферы

# --------------------------------
# 2. Функция для расчёта сегмента
# --------------------------------
def calculate_sphere_segment(R, D):
    """
    Параметры:
    R (float): радиус сферы
    D (float): диаметр основания шарового сегмента
    
    Возвращает (d, h, delta):
    d     = расстояние от центра сферы до плоскости сечения
    h     = высота шарового сегмента
    delta = глубина шлифовки (R - h)
    """
    # Расстояние от центра сферы до плоскости сечения
    d = math.sqrt(R**2 - (D / 2)**2)
    # Высота шарового сегмента
    h = R - d
    # Глубина шлифовки (сколько "снято")
    delta = R - h  # то же самое, что d
    
    return d, h, delta

# --------------------------------
# 3. Функция для визуализации
# --------------------------------
def visualize_multilayer_sphere(R, layer_thicknesses, D, d):
    """
    Рисует:
      - Многослойную сферу в виде концентрических окружностей.
      - Хорду (диаметр D) на высоте y = d (от центра).
      - Заштрихованный шаровой сегмент над этой хордой.
    
    Параметры:
    R (float): общий радиус сферы
    layer_thicknesses (list of float): толщина каждого слоя
    D (float): диаметр сечения
    d (float): расстояние от центра до плоскости сечения
    """
    # --------------------------
    # Вычислим границы слоёв.
    # Предположим, что layer_thicknesses = [t1, t2, t3, t4]
    # и R = t1 + t2 + t3 + t4
    # Тогда радиусы границ слоёв (от центра к краю):
    #
    # r0 = 0
    # r1 = t1
    # r2 = t1 + t2
    # r3 = t1 + t2 + t3
    # r4 = t1 + t2 + t3 + t4 = R
    # --------------------------
    boundaries = [0.0]
    cum_sum = 0.0
    for t in layer_thicknesses:
        cum_sum += t
        boundaries.append(cum_sum)
    
    # Для построения окружностей подготовим "угол" от 0 до 2π
    theta = np.linspace(0, 2 * np.pi, 360)
    
    # Координата y, на которой расположена хорда
    # (считаем, что центр сферы в (0,0),
    #  тогда плоскость сечения параллельна оси X)
    chord_y = d  # это же c в формулах
    
    # --------------------------
    # Создаём фигуру
    # --------------------------
    plt.figure(figsize=(6, 6))
    plt.title("Многослойная сфера (поперечное сечение)")
    
    # 1) Рисуем слои — каждый слой будет своим цветом
    colors = ["#FFC0CB", "#ADD8E6", "#90EE90", "#FFFFE0"]  # Можно менять по вкусу
    
    for i in range(1, len(boundaries)):
        # Наружный радиус текущего слоя
        r_outer = boundaries[i]
        # Внутренний радиус текущего слоя
        r_inner = boundaries[i-1]
        
        # Координаты окружности для внешней границы
        x_outer = r_outer * np.cos(theta)
        y_outer = r_outer * np.sin(theta)
        
        # Чтобы "закрасить" слой, заполненный между r_inner и r_outer:
        # - рисуем внешнюю границу
        plt.plot(x_outer, y_outer, color='k', linewidth=1)
        
        # Для заливки между r_inner и r_outer воспользуемся fill_between:
        #  * Но fill_between работает по вертикали, поэтому проще
        #    построить полукольца вдоль массива x и y. 
        # Чтобы не усложнять, используем "математическую" заливку
        # (см. чуть ниже после отрисовки хорды и сегмента).
        
        # Для аккуратного закрашивания сразу всех слоёв
        # удобнее использовать fill_between внутри "лучей".
        # Но можно пойти проще и нарисовать каждый слой как окружность
        # с помощью plt.fill(x, y, ...)
        
        # Сгенерируем "заполненные" координаты для внешней и внутренней границ
        # (берём много точек, чтобы форма была гладкой)
        points = 180  # количество точек для отрисовки
        angles = np.linspace(0, 2*np.pi, points)
        x_outer_fill = r_outer * np.cos(angles)
        y_outer_fill = r_outer * np.sin(angles)
        x_inner_fill = r_inner * np.cos(angles)
        y_inner_fill = r_inner * np.sin(angles)
        
        # Формируем контур слоя:
        # - сначала внешний контур по часовой стрелке,
        # - потом внутренний контур в обратном порядке
        x_layer = np.concatenate([x_outer_fill, x_inner_fill[::-1]])
        y_layer = np.concatenate([y_outer_fill, y_inner_fill[::-1]])
        
        plt.fill(x_layer, y_layer, color=colors[i-1], alpha=0.6)
    
    # 2) Рисуем хорду (сечение диаметром D)
    half_d = D / 2
    plt.plot([-half_d, half_d], [chord_y, chord_y],
             'r-', linewidth=2, label="Диаметр основания (D)")
    
    # 3) Заштриховываем сегмент над хордой
    #    Для этого выберем точечно x в диапазоне [-D/2, D/2] и
    #    найдём верхнюю полуокружность (y >= 0 для всей сферы? или
    #    вообще вся окружность?). Так как сегмент "над" хордой,
    #    предполагаем, что срез располагается с "верхней" стороны сферы.
    
    x_fill_segment = np.linspace(-half_d, half_d, 200)
    # Верхняя дуга окружности (полного радиуса R)
    y_fill_upper = np.sqrt(R**2 - x_fill_segment**2)
    # Всё, что выше chord_y и ниже y_fill_upper
    # (если chord_y>0, значит срез где-то выше оси)
    
    # Однако chord_y может быть положительным или отрицательным,
    # но "сегмент" — это часть от chord_y до "верха".
    # На всякий случай возьмём min/chord_y, но логика такова:
    #   segment_y_low = chord_y
    #   segment_y_high = y_fill_upper
    segment_y_low = np.full_like(x_fill_segment, chord_y)
    segment_y_high = y_fill_upper
    
    plt.fill_between(x_fill_segment, segment_y_low, segment_y_high,
                     color='gray', alpha=0.3, label="Шаровой сегмент")
    
    # Отметим центр сферы (0, 0)
    plt.plot(0, 0, 'ko', label="Центр сферы")
    
    # Настраиваем оси
    plt.axis('equal')
    plt.grid(True)
    
    # Выставим границы, чтобы сфера полностью помещалась
    pad = 0.1 * R
    plt.xlim(-R - pad, R + pad)
    plt.ylim(-R - pad, R + pad)
    
    plt.legend()
    plt.show()

# --------------------------------
# 4. Основная "точка входа"
# --------------------------------
def main():
    print("Многослойная сфера из 4 слоёв.")
    print(f"Слои (толщины, мм): {layer_thicknesses}")
    print(f"Итоговый радиус сферы R = {R:.2f} мм\n")
    
    # Ввод диаметра
    D = float(input("Введите диаметр D основания шарового сегмента (мм): "))
    
    # Проверка, что D <= 2R (иначе плоскость не пересечёт сферу)
    if D > 2*R:
        print("Ошибка: диаметр D больше диаметра самой сферы!")
        return
    
    # Считаем основные параметры
    d, h, delta = calculate_sphere_segment(R, D)
    
    # Выводим результаты
    print(f"Расстояние d от центра до плоскости сечения: {d:.4f} мм")
    print(f"Высота шарового сегмента h: {h:.4f} мм")
    print(f"Глубина шлифовки (Δ = R - h): {delta:.4f} мм")
    
    # Визуализируем
    visualize_multilayer_sphere(R, layer_thicknesses, D, d)

if __name__ == "__main__":
    main()
```

```python
import math
import matplotlib.pyplot as plt
import numpy as np

# --------------------------------
# 1. Задаём толщины 4 слоёв (в микрометрах).
#    Ниже пример: 4 слоя по 5000, 3000, 2000, 4000 мкм.
#    Суммарно дают радиус 14 000 мкм (14 мм).
# --------------------------------
layer_thicknesses = [250.0, 95.0, 40.0, 35.0, 40.0]  # мкм
R = sum(layer_thicknesses)  # радиус сферы в мкм

# --------------------------------
# 2. Функция для расчёта сегмента
# --------------------------------
def calculate_sphere_segment(R, D):
    """
    Параметры:
    R (float): радиус сферы (мкм)
    D (float): диаметр основания шарового сегмента (мкм)
    
    Возвращает (d, h, delta, V_removed):
    d         = расстояние от центра сферы до плоскости сечения (мкм)
    h         = высота шарового сегмента (мкм)
    delta     = глубина шлифовки (R - h) (мкм)
    V_removed = объём снятого материала (куб. мкм)
    """
    # Расстояние от центра сферы до плоскости среза
    d = math.sqrt(R**2 - (D / 2)**2)
    # Высота шарового сегмента
    h = R - d
    # Глубина шлифовки (сколько "срезано")
    delta = R - h  # то же самое, что d
    
    # Объём снятого сегмента (шаровой сегмент высотой h)
    # Формула V = (π * h^2 * (3R - h)) / 3
    V_removed = (math.pi * (h**2) * (3*R - h)) / 3.0
    
    return d, h, delta, V_removed

# --------------------------------
# 3. Функция для визуализации
# --------------------------------
def visualize_multilayer_sphere(R, layer_thicknesses, D, d):
    """
    Рисует в 2D (поперечное сечение):
      - Многослойную сферу (концентрические окружности).
      - Хорду (диаметр D) на высоте y = d.
      - Заштрихованный шаровой сегмент над этой хордой.
    
    Параметры:
    R (float): общий радиус сферы (мкм)
    layer_thicknesses (list of float): толщина каждого слоя (мкм)
    D (float): диаметр сечения (мкм)
    d (float): расстояние от центра до плоскости сечения (мкм)
    """
    # Радиусы «границ» слоёв: накапливаем сумму толщин
    boundaries = [0.0]
    cum_sum = 0.0
    for t in layer_thicknesses:
        cum_sum += t
        boundaries.append(cum_sum)
    
    # Подготовка углов для рисования окружностей
    theta = np.linspace(0, 2*np.pi, 360)
    
    # Координата y, где лежит хорда
    chord_y = d
    
    # Создаём окно
    plt.figure(figsize=(6, 6))
    plt.title("Многослойная сфера (поперечное сечение, в микрометрах)")
    
    # 1) Отрисовка слоёв
    # Подберём цвета слоёв (можно менять по желанию)
    colors = ["#FFDF00", "#db7093", "#494d4e", "#6495ed", "#3b3d3e"] 
    
    for i in range(1, len(boundaries)):
        r_outer = boundaries[i]   # внешний радиус данного слоя
        r_inner = boundaries[i-1] # внутренний радиус данного слоя
        
        # Генерируем координаты внешней границы
        x_outer = r_outer * np.cos(theta)
        y_outer = r_outer * np.sin(theta)
        
        # Рисуем внешнюю окружность слоя
        plt.plot(x_outer, y_outer, color='k', linewidth=1)
        
        # Для заливки слоя между r_inner и r_outer
        angles = np.linspace(0, 2*np.pi, 360)
        x_outer_fill = r_outer * np.cos(angles)
        y_outer_fill = r_outer * np.sin(angles)
        x_inner_fill = r_inner * np.cos(angles)
        y_inner_fill = r_inner * np.sin(angles)
        
        # Формируем замкнутый контур слоя
        x_layer = np.concatenate([x_outer_fill, x_inner_fill[::-1]])
        y_layer = np.concatenate([y_outer_fill, y_inner_fill[::-1]])
        
        plt.fill(x_layer, y_layer, color=colors[i-1], alpha=0.6)
    
    # 2) Рисуем хорду (красная линия)
    half_d = D / 2
    plt.plot([-half_d, half_d], [chord_y, chord_y],
             'r-', linewidth=2, label="Диаметр среза (D)")
    
    # 3) Заштриховываем шаровой сегмент, который "срезан" сверху
    x_fill_segment = np.linspace(-half_d, half_d, 200)
    y_fill_upper = np.sqrt(R**2 - x_fill_segment**2)  # верхняя часть сферы
    y_fill_lower = np.full_like(x_fill_segment, chord_y)  # плоскость среза
    
    plt.fill_between(x_fill_segment, y_fill_lower, y_fill_upper,
                     color='gray', alpha=0.7, label="Снятый сегмент")
    
    # Отметим центр сферы
    plt.plot(0, 0, 'ko', label="Центр сферы (0, 0)")
    
    # Настраиваем оси
    plt.axis('equal')
    plt.grid(True)
    pad = 0.1 * R
    plt.xlim(-R - pad, R + pad)
    plt.ylim(-R - pad, R + pad)
    
    plt.legend()
    plt.show()

# --------------------------------
# 4. Основная функция (точка входа)
# --------------------------------
def main():
    print("Многослойная сфера из 4 слоёв (все размеры — в микрометрах).")
    print(f"Слои (толщины, мкм): {layer_thicknesses}")
    print(f"Суммарный радиус сферы R = {R:.2f} мкм\n")
    
    # Ввод диаметра среза
    D = float(input("Введите диаметр D основания шарового сегмента (мкм): "))
    
    # Проверка, чтобы хорда не была больше диаметра сферы
    if D > 2*R:
        print("Ошибка: введённый D превышает диаметр сферы! Срез невозможен.")
        return
    
    # Считаем основные параметры
    d, h, delta, V_removed = calculate_sphere_segment(R, D)
    
    # Выводим результаты
    print("\n--- Результаты ---")
    print(f"Расстояние d от центра до плоскости среза: {d:.4f} мкм")
    print(f"Высота шарового сегмента h:              {h:.4f} мкм")
    print(f"Глубина шлифовки (Δ = R - h):            {delta:.4f} мкм")
    print(f"Объём снятого материала (V_removed):     {V_removed:.4f} куб. мкм")
    
    # Визуализируем
    visualize_multilayer_sphere(R, layer_thicknesses, D, d)

if __name__ == "__main__":
    main()
```

```python
import math
import matplotlib.pyplot as plt
import numpy as np

# -------------------------------------------------
# 1. Задаём ПЯТЬ слоёв (в микрометрах) и их названия
# -------------------------------------------------
layer_thicknesses = [250.0, 95.0, 40.0, 35.0, 40.0]  # мкм
layer_names       = ["1", "2", "3", "4", "5"]

# Суммарный радиус сферы (мкм)
R = sum(layer_thicknesses)

# -------------------------------------------------
# 2. Функция для вычисления основных параметров
# -------------------------------------------------
def calculate_sphere_segment(R, D):
    """
    Параметры:
      R (float): радиус сферы, мкм
      D (float): диаметр среза шарового сегмента, мкм
    
    Возвращает:
      d         = расстояние от центра сферы до плоскости сечения (мкм)
      h         = высота шарового сегмента (мкм)
      delta     = глубина шлифовки (R - h) (мкм)
      V_removed = объём снятого материала (куб. мкм)
    """
    # Расстояние от центра сферы до плоскости
    d = math.sqrt(R**2 - (D/2)**2)
    
    # Высота шарового сегмента
    h = R - d
    
    # Глубина шлифовки (то есть сколько "убрали" сверху)
    delta = R - h  # то же, что d
    
    # Объём снятого сегмента по формуле: V = (π*h²*(3R - h)) / 3
    V_removed = (math.pi * (h**2) * (3*R - h)) / 3.0
    
    return d, h, delta, V_removed

# -------------------------------------------------
# 3. Функция для визуализации многослойной сферы
# -------------------------------------------------
def visualize_multilayer_sphere(R, layer_thicknesses, layer_names, D, d):
    """
    Рисует в 2D поперечное сечение многослойной сферы:
      - 5 концентрических окружностей (каждый слой своим цветом),
      - Хорду (диаметр D) на высоте y = d,
      - Заштрихованный шаровой сегмент выше этой хорды,
      - Подписи с названиями слоёв (1-UO2, 2-Buffer, и т.д.).
    
    Параметры:
      R                 (float): общий радиус сферы (мкм)
      layer_thicknesses (list) : толщины слоёв (мкм)
      layer_names       (list) : названия слоёв
      D                 (float): диаметр основания сегмента (мкм)
      d                 (float): расстояние от центра до плоскости среза (мкм)
    """
    # Найдём границы каждого слоя от центра наружу
    # Пример: если layer_thicknesses = [5000,3000,2000,4000,5000]
    # то границы будут [0, 5000, 8000, 10000, 14000, 19000]
    boundaries = [0.0]
    cum_sum = 0.0
    for t in layer_thicknesses:
        cum_sum += t
        boundaries.append(cum_sum)
    
    # Подготовим массив углов для окружностей
    theta = np.linspace(0, 2*np.pi, 360)
    
    # Хорда на высоте chord_y = d
    chord_y = d
    
    plt.figure(figsize=(7, 7))
    plt.title("Многослойная сфера (сечение) в микрометрах")
    
    # 1) Отрисовка слоёв в виде концентрических колец
    # Зададим 5 цветов — по одному на каждый слой
    colors = ["#FFDF00", "#db7093", "#494d4e", "#6495ed", "#3b3d3e"]
    
    for i in range(len(layer_thicknesses)):
        r_inner = boundaries[i]     # внутренний радиус i-го слоя
        r_outer = boundaries[i+1]   # внешний радиус i-го слоя
        color   = colors[i]         # цвет для i-го слоя
        name    = layer_names[i]    # название слоя, напр. "1-UO2"
        
        # Нарисуем внешнюю границу (окружность) радиуса r_outer
        x_outer = r_outer * np.cos(theta)
        y_outer = r_outer * np.sin(theta)
        plt.plot(x_outer, y_outer, color='k', linewidth=1)
        
        # Заполним цветом кольцо между r_inner и r_outer
        # Сгенерируем координаты для заливки
        angles = np.linspace(0, 2*np.pi, 360)
        x_outer_fill = r_outer * np.cos(angles)
        y_outer_fill = r_outer * np.sin(angles)
        x_inner_fill = r_inner * np.cos(angles)
        y_inner_fill = r_inner * np.sin(angles)
        
        # Формируем замкнутый контур слоя
        x_layer = np.concatenate([x_outer_fill, x_inner_fill[::-1]])
        y_layer = np.concatenate([y_outer_fill, y_inner_fill[::-1]])
        
        plt.fill(x_layer, y_layer, color=color, alpha=0.6)
        
        # Добавим текстовое обозначение слоя
        # Возьмём радиус "по середине" и поставим текст справа от центра
        r_mid = (r_inner + r_outer) / 2
        plt.text(r_mid, 0, name, fontsize=9,
                 ha='left', va='center',
                 color='black',
                 rotation=0)
    
    # 2) Хорда (красная линия) — диаметр среза
    half_d = D / 2.0
    plt.plot([-half_d, half_d], [chord_y, chord_y],
             'r-', linewidth=2, label="Диаметр сегмента (D)")
    
    # 3) Заштриховываем "срезанный" шаровой сегмент сверху
    x_fill_segment = np.linspace(-half_d, half_d, 200)
    y_fill_upper = np.sqrt(R**2 - x_fill_segment**2)  # верхняя дуга сферы
    y_fill_lower = np.full_like(x_fill_segment, chord_y)
    
    plt.fill_between(x_fill_segment, y_fill_lower, y_fill_upper,
                     color='gray', alpha=0.3, label="Снятый сегмент")
    
    # Отметим центр сферы (0,0)
    plt.plot(0, 0, 'ko', label="Центр сферы (0,0)")
    
    plt.axis('equal')
    plt.grid(True)
    
    # Зададим границы
    pad = 0.1 * R
    plt.xlim(-R - pad, R + pad)
    plt.ylim(-R - pad, R + pad)
    
    plt.legend()
    plt.show()

# -------------------------------------------------
# 4. Главная функция (точка входа)
# -------------------------------------------------
def main():
    print("Многослойная сфера из 5 слоёв (в микрометрах).")
    print(f"Слои (толщины, мкм): {layer_thicknesses}")
    print(f"Названия слоёв: {layer_names}")
    print(f"Суммарный радиус сферы R = {R:.2f} мкм\n")
    
    # Спросим диаметр среза
    D = float(input("Введите диаметр D шлифа шарового сегмента (мкм): "))
    
    # Проверка: нельзя, чтобы D > 2*R (тогда срез не пересечёт сферу корректно)
    if D > 2*R:
        print("Ошибка: введённый D больше диаметра сферы! Срез невозможен.")
        return
    
    # Считаем основные параметры
    d, h, delta, V_removed = calculate_sphere_segment(R, D)
    
    # Вывод результатов
    print("\n--- Результаты расчётов ---")
    print(f"Расстояние d от центра до плоскости среза: {d:.4f} мкм")
    print(f"Высота шарового сегмента h:              {h:.4f} мкм")
    print(f"Глубина шлифовки (Δ = R - h):            {delta:.4f} мкм")
    print(f"Объём снятого материала V_removed:       {V_removed:.4f} куб. мкм")
    
    # Визуализация
    visualize_multilayer_sphere(R, layer_thicknesses, layer_names, D, d)

if __name__ == "__main__":
    main()
```
